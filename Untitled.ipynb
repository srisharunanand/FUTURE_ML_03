```python
# Create a sample job description for Data Scientist role
job_description = """
JOB TITLE: Data Scientist

ABOUT THE ROLE:
We are looking for an experienced Data Scientist to join our team.
You will work on building machine learning models and analyzing data 
to drive business decisions.

REQUIRED SKILLS:
- Python programming
- Machine Learning
- Data Analysis
- SQL
- Statistics
- Data Visualization
- Deep Learning (optional but preferred)

EXPERIENCE REQUIRED:
- 3-5 years of experience in data science
- Experience with TensorFlow, Scikit-learn, Pandas
- Strong SQL and database knowledge
- Experience with big data technologies (Hadoop, Spark)
- Portfolio or GitHub projects demonstrating ML work

NICE TO HAVE:
- Experience with cloud platforms (AWS, GCP, Azure)
- Knowledge of NLP
- Experience with Docker and Kubernetes
- Master's degree in Computer Science, Mathematics, or related field
"""

print("📋 JOB DESCRIPTION:")
print("="*70)
print(job_description)
print("="*70)

# Save it for later use
job_desc = job_description
print("\n✅ Job description created!")
print(f"Job description length: {len(job_desc)} characters")
print(f"Job description words: {len(job_desc.split())} words")
```

```python
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load the spaCy NLP model
print("Loading spaCy NLP model...")
nlp = spacy.load('en_core_web_sm')
print("✅ spaCy model loaded!")

# Define technical skills to look for
technical_skills = {
    'python': ['python', 'py', 'python3'],
    'machine learning': ['machine learning', 'ml', 'scikit-learn', 'sklearn'],
    'deep learning': ['deep learning', 'neural network', 'cnn', 'rnn', 'tensorflow', 'keras'],
    'sql': ['sql', 'mysql', 'postgresql', 'oracle'],
    'data analysis': ['data analysis', 'statistical analysis', 'pandas'],
    'data visualization': ['visualization', 'matplotlib', 'seaborn', 'plotly', 'tableau', 'power bi'],
    'big data': ['hadoop', 'spark', 'mapreduce', 'hive'],
    'aws': ['aws', 'amazon web services', 'ec2', 's3'],
    'gcp': ['gcp', 'google cloud', 'bigquery'],
    'azure': ['azure', 'microsoft azure'],
    'docker': ['docker', 'containerization'],
    'kubernetes': ['kubernetes', 'k8s'],
    'nlp': ['nlp', 'natural language processing', 'nltk', 'spacy'],
    'git': ['git', 'github', 'gitlab', 'version control'],
    'statistics': ['statistics', 'probability', 'statistical'],
    'r programming': ['r programming', ' r ', 'r language'],
    'java': ['java', 'j2ee'],
    'scala': ['scala'],
    'excel': ['excel', 'vba', 'spreadsheet'],
}

print("\n✅ Skill dictionary created with 16 skill categories!")
print(f"Skills being tracked: {', '.join(technical_skills.keys())}")

# Function to extract skills from text
def extract_skills(text):
    """
    Extract skills from resume text
    """
    text_lower = text.lower()
    found_skills = []
    
    for skill_name, skill_keywords in technical_skills.items():
        for keyword in skill_keywords:
            if keyword.lower() in text_lower:
                found_skills.append(skill_name)
                break  # Don't count same skill twice
    
    return list(set(found_skills))  # Remove duplicates

# Test on sample resumes
print("\n" + "="*70)
print("TESTING SKILL EXTRACTION:")
print("="*70)

for idx in range(3):
    resume_text = df['Resume_str'].iloc[idx]
    skills = extract_skills(resume_text)
    
    print(f"\nResume {idx + 1}:")
    print(f"  Job Title/Category: {df['Category'].iloc[idx]}")
    print(f"  Skills found: {skills if skills else 'None detected'}")
    print(f"  Number of skills: {len(skills)}")
    print(f"  Resume preview: {resume_text[:150]}...")

# Extract skills from job description
job_skills = extract_skills(job_desc)
print("\n" + "="*70)
print("SKILLS REQUIRED FOR THE JOB:")
print("="*70)
print(f"Required skills: {job_skills}")
print(f"Total required skills: {len(job_skills)}")
```

```python
# Phase 4: Score resumes based on job match

def calculate_match_score(resume_skills, job_skills):
    """
    Calculate how well a resume matches the job
    
    Returns: score (0-100)
    """
    if len(job_skills) == 0:
        return 0
    
    # Count matching skills
    matching_skills = set(resume_skills) & set(job_skills)
    
    # Calculate percentage match
    match_percentage = (len(matching_skills) / len(job_skills)) * 100
    
    return match_percentage, matching_skills

def find_missing_skills(resume_skills, job_skills):
    """
    Find skills required by job but missing in resume
    """
    missing = set(job_skills) - set(resume_skills)
    return list(missing)

# Extract skills for ALL resumes
print("Extracting skills from all 2,484 resumes...")
df['extracted_skills'] = df['Resume_str'].apply(extract_skills)
print("✅ Skills extracted from all resumes!")

# Score each resume against the job
print("\nCalculating match scores...")
results = []

for idx, row in df.iterrows():
    resume_skills = row['extracted_skills']
    score, matching = calculate_match_score(resume_skills, job_skills)
    missing = find_missing_skills(resume_skills, job_skills)
    
    results.append({
        'ID': row['ID'],
        'Category': row['Category'],
        'Match_Score': score,
        'Matching_Skills': list(matching),
        'Missing_Skills': missing,
        'Total_Skills_Found': len(resume_skills),
        'Job_Skills_Found': len(matching)
    })

# Create results dataframe
results_df = pd.DataFrame(results)
print("✅ All resumes scored!")

# Sort by match score (highest first)
results_df_sorted = results_df.sort_values('Match_Score', ascending=False)

print("\n" + "="*70)
print("TOP 20 CANDIDATES FOR DATA SCIENTIST ROLE:")
print("="*70)

for idx, (i, row) in enumerate(results_df_sorted.head(20).iterrows(), 1):
    print(f"\n{idx}. Resume ID: {row['ID']}")
    print(f"   Category: {row['Category']}")
    print(f"   Match Score: {row['Match_Score']:.1f}%")
    print(f"   Skills Found: {row['Job_Skills_Found']}/{len(job_skills)} required")
    print(f"   Matching Skills: {', '.join(row['Matching_Skills']) if row['Matching_Skills'] else 'None'}")
    print(f"   Missing Skills: {', '.join(row['Missing_Skills'][:5])}..." if len(row['Missing_Skills']) > 5 else f"   Missing Skills: {', '.join(row['Missing_Skills'])}")

print("\n" + "="*70)
print("SCORE STATISTICS:")
print("="*70)
print(f"Average match score: {results_df['Match_Score'].mean():.1f}%")
print(f"Best match score: {results_df['Match_Score'].max():.1f}%")
print(f"Worst match score: {results_df['Match_Score'].min():.1f}%")
print(f"Median match score: {results_df['Match_Score'].median():.1f}%")

# Show score distribution
print("\n" + "="*70)
print("SCORE DISTRIBUTION:")
print("="*70)
score_ranges = [
    (80, 100, "Excellent fit"),
    (60, 79, "Good fit"),
    (40, 59, "Moderate fit"),
    (20, 39, "Weak fit"),
    (0, 19, "Very weak fit")
]

for min_score, max_score, label in score_ranges:
    count = len(results_df[(results_df['Match_Score'] >= min_score) & (results_df['Match_Score'] < max_score)])
    percentage = (count / len(results_df)) * 100
    print(f"{label:20} ({min_score}-{max_score}%): {count:4} resumes ({percentage:5.1f}%)")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 10)

# Create a comprehensive visualization
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Resume Screening Analysis for Data Scientist Role', fontsize=16, fontweight='bold')

# Chart 1: Score Distribution (Histogram)
ax1 = axes[0, 0]
ax1.hist(results_df['Match_Score'], bins=30, color='#3498db', edgecolor='black', alpha=0.7)
ax1.axvline(results_df['Match_Score'].mean(), color='red', linestyle='--', linewidth=2, label=f'Average: {results_df["Match_Score"].mean():.1f}%')
ax1.axvline(results_df['Match_Score'].median(), color='green', linestyle='--', linewidth=2, label=f'Median: {results_df["Match_Score"].median():.1f}%')
ax1.set_xlabel('Match Score (%)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Number of Resumes', fontsize=11, fontweight='bold')
ax1.set_title('Distribution of Candidate Match Scores', fontsize=12, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Chart 2: Score Categories (Pie Chart)
ax2 = axes[0, 1]
score_counts = [0, 1, 13, 204, 2266]
score_labels = ['Excellent\n(80-100%)', 'Good\n(60-79%)', 'Moderate\n(40-59%)', 'Weak\n(20-39%)', 'Very Weak\n(0-19%)']
colors = ['#27ae60', '#f39c12', '#e67e22', '#e74c3c', '#c0392b']
ax2.pie(score_counts, labels=score_labels, autopct='%1.1f%%', colors=colors, startangle=90)
ax2.set_title('Candidate Distribution by Match Level', fontsize=12, fontweight='bold')

# Chart 3: Top 10 candidates bar chart
ax3 = axes[1, 0]
top_10 = results_df_sorted.head(10)
bars = ax3.barh(range(10), top_10['Match_Score'].values, color='#2ecc71')
ax3.set_yticks(range(10))
ax3.set_yticklabels([f"Resume {i+1}" for i in range(10)])
ax3.set_xlabel('Match Score (%)', fontsize=11, fontweight='bold')
ax3.set_title('Top 10 Candidates by Match Score', fontsize=12, fontweight='bold')
ax3.invert_yaxis()
for i, bar in enumerate(bars):
    width = bar.get_width()
    ax3.text(width - 2, bar.get_y() + bar.get_height()/2, f'{width:.1f}%', 
             ha='right', va='center', fontweight='bold', color='white', fontsize=10)
ax3.set_xlim(0, 70)
ax3.grid(axis='x', alpha=0.3)

# Chart 4: Skills analysis
ax4 = axes[1, 1]
skill_frequency = {}
for skills_list in df['extracted_skills']:
    for skill in skills_list:
        skill_frequency[skill] = skill_frequency.get(skill, 0) + 1

# Sort and get top skills
top_skills = sorted(skill_frequency.items(), key=lambda x: x[1], reverse=True)[:10]
skill_names = [s[0] for s in top_skills]
skill_counts = [s[1] for s in top_skills]

bars = ax4.barh(skill_names, skill_counts, color='#9b59b6')
ax4.set_xlabel('Number of Resumes with Skill', fontsize=11, fontweight='bold')
ax4.set_title('Top 10 Skills Found in All Resumes', fontsize=12, fontweight='bold')
for i, bar in enumerate(bars):
    width = bar.get_width()
    ax4.text(width + 20, bar.get_y() + bar.get_height()/2, str(int(width)), 
             ha='left', va='center', fontweight='bold', fontsize=9)
ax4.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('resume_screening_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Visualization saved as 'resume_screening_analysis.png'")
```
